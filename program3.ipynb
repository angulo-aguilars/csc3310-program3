## Approach Description:
This program focuses on identifying the longest sequence of matching characters that end at specific indices in two strings a and b. State (i,j) represents the length of the common substring ending at a[i] and b[j]

- For backtracking: a recursive function where if a[i] is equal to b[j] then the result is 1 + the result of the previous pair (i-1, j-1)

- For memoization: The result of each (i,j) pair is stored in a 2D array, in order to avoid the exponential redundant calls of brute force.

- For eager dynamic programming: Recursion is replaced with nested loops to fill the 2D array from the bottom up. This ensures that every subproblem is solved exactly once in a predictable order

Backtracking Pseudocode

```
computeMaxAt(i, j, str_a, str_b):
    if i < 0 or j < 0
        then return 0
    if str_a[i] == str_b[j]
        return 1 + computeMaxAt(i-1, j-1, str_a, str_b)
    else
        return 0

backtrackingSearch(a, b):
    max_length = 0 
    for i from 0 to length(a)-1
        for j from 0 to length(b)-1
            curr = computeMaxAt(i, j, a, b)
            if curr > max_length
                max_length = curr
    return max_length
```


Memoization Pseudocode

```
memo_table[length(a)][length(b)] initialized with -1
memoizedMaxAt(i, j, a, b):
    if i < 0 OR j < 0 
        return 0
    if memo_table[i][j] != -1 
        return memo_table[i][j]
    if a[i] == b[j]
        memo_table[i][j] = 1 + memoizedMaxAt(i-1, j-1, a, b)
    else 
        memo_table[i][j] = 0
    return memo_table[i][j]
```

Eager Dynamic Programming Pseudocode

```
eagerDP(a, b):
    n = length(a), m = length(b)
    table[n+1][m+1] initialized to 0
    max_val = 0
    end_pos = 0
    
    for i from 1 to n
        for j from 1 to m
            if a[i-1] == b[j-1]
                table[i][j] = table[i-1][j-1] + 1
                table[i][j] > max_val
                    max_val = table[i][j]
                    end_pos = i
            else
                table[i][j] = 0
    return substring of a from (end_pos - max_val) to end_pos
```   

````python
#implementation
class MaximalSubstring:
    def __init__(self, a, b):
        self.a = a
        self.b = b
        self.memo = [[-1 for _ in range(len(b))] for _ in range(len(a))]

    def backtrack_recursive(self, i, j):
        if i < 0 or j < 0:
            return 0
        if self.a[i] == self.b[j]:
            return 1 + self.backtrack_recursive(i - 1, j - 1)
        return 0

    def solve_backtracking(self):
        max_len = 0
        for i in range(len(self.a)):
            for j in range(len(self.b)):
                max_len = max(max_len, self.backtrack_recursive(i, j))
        return max_len

    def memoized_recursive(self, i, j):
        if i < 0 or j < 0:
            return 0
        if self.memo[i][j] != -1:
            return self.memo[i][j]
        
        if self.a[i] == self.b[j]:
            self.memo[i][j] = 1 + self.memoized_recursive(i - 1, j - 1)
        else:
            self.memo[i][j] = 0
        return self.memo[i][j]

    def solve_memoized(self):
        max_len = 0
        for i in range(len(self.a)):
            for j in range(len(self.b)):
                max_len = max(max_len, self.memoized_recursive(i, j))
        return max_len

    def solve_eager(self):
        n, m = len(self.a), len(self.b)
        table = [[0] * (m + 1) for _ in range(n + 1)]
        max_len = 0
        end_idx = 0

        for i in range(1, n + 1):
            for j in range(1, m + 1):
                if self.a[i-1] == self.b[j-1]:
                    table[i][j] = table[i-1][j-1] + 1
                    if table[i][j] > max_len:
                        max_len = table[i][j]
                        end_idx = i
                else:
                    table[i][j] = 0
        return self.a[end_idx - max_len : end_idx]
````

````python
#running code
solver = MaximalSubstring("banana", "orange")
print(solver.solve_eager()) 
````

# Benchmarking

````python
import time
import random
import string
import gc



def get_case(length, scenario):
    if scenario == "Best":
        return "a" * length, "x" * length
    elif scenario == "Worst":
        return "a" * length, "a" * length
    elif scenario == "Average":
        a = ''.join(random.choices(string.ascii_lowercase, k=length))
        b = ''.join(random.choices(string.ascii_lowercase, k=length))
        return a, b


def benchmark(algorithm):
    lengths = [10, 100, 1000, 5000, 10000]
    scenarios = ["Best", "Average", "Worst"]
    
    def get_iterations(n):
        if n <= 100: return 500
        if n <= 1000: return 50
        if n <= 5000: return 5
        return 2

    print(f"| Length | Best Case (s) | Avg Case (s) | Worst Case (s) | Iterations |")
    print(f"| :----- | :-------------| :----------- | :------------- | :--------- |")

    for n in lengths:
        results = {}
        iterations = get_iterations(n)
        
        for scenario in scenarios:
            a, b = get_case(n, scenario)
            
            solver = MaximalSubstring(a, b)

            alg = getattr(solver, algorithm)
            
            gc.disable() 
            
            start_time = time.perf_counter()
            
            for _ in range(iterations):
                alg()
                
            end_time = time.perf_counter()
            
            gc.enable() 
                        
            total_time = end_time - start_time
            avg_time = total_time / iterations
            results[scenario] = avg_time
            
        print(f"| {n:<6} | {results['Best']:<13.6f} | {results['Average']:<12.6f} | {results['Worst']:<14.6f} | {iterations:<10} |")
  
````

````python
benchmark("solve_eager")
````

````python
benchmark("solve_backtracking")
````
