## Approach Description:
This program focuses on identifying the longest sequence of matching characters that end at specific indices in two strings a and b. State (i,j) represents the length of the common substring ending at a[i] and b[j]

- For backtracking: a recursive function where if a[i] is equal to b[j] then the result is 1 + the result of the previous pair (i-1, j-1)

- For memoization: The result of each (i,j) pair is stored in a 2D array, in order to avoid the exponential redundant calls of brute force.

- For eager dynamic programming: Recursion is replaced with nested loops to fill the 2D array from the bottom up. This ensures that every subproblem is solved exactly once in a predictable order

Backtracking Pseudocode
computeMaxAt(i, j, str_a, str_b):
    if i < 0 or j < 0
        then return 0
    if str_a[i] == str_b[j]
        return 1 + computeMaxAt(i-1, j-1, str_a, str_b)
    else
        return 0

backtrackingSearch(a, b):
    max_length = 0 
    for i from 0 to length(a)-1
        for j from 0 to length(b)-1
            curr = computeMaxAt(i, j, a, b)
            if curr > max_length
                max_length = curr
    return max_length


Memoization Pseudocode
memo_table[length(a)][length(b)] initialized with -1

memoizedMaxAt(i, j, a, b):
    if i < 0 OR j < 0 
        return 0
    if memo_table[i][j] != -1 
        return memo_table[i][j]
    if a[i] == b[j]
        memo_table[i][j] = 1 + memoizedMaxAt(i-1, j-1, a, b)
    else 
        memo_table[i][j] = 0
    return memo_table[i][j]

Eager Dynamic Programming Pseudocode
eagerDP(a, b):
    n = length(a), m = length(b)
    table[n+1][m+1] initialized to 0
    max_val = 0
    end_pos = 0
    
    for i from 1 to n
        FOR j from 1 to m
            if a[i-1] == b[j-1]
                table[i][j] = table[i-1][j-1] + 1
                table[i][j] > max_val
                    max_val = table[i][j]
                    end_pos = i
            else
                table[i][j] = 0
    return substring of a from (end_pos - max_val) to end_pos
